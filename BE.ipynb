# Top Interview Questions

These **28 core backend engineering questions** account for approximately 80% of what interviewers repeatedly test across Mid-to-Senior Backend, Systems, and API interviews. They are grouped by functional domain along with the exact core concepts you are expected to articulate.

---

### 1. Networking, Protocols & API Design

* **1. What happens under the hood when a client makes an HTTPS request to a server?**
* *Concepts:* DNS resolution, TCP 3-way handshake (`SYN`, `SYN-ACK`, `ACK`), TLS/SSL handshake (key exchange and symmetric encryption), and HTTP layer request/response cycle.


* **2. How do REST, gRPC, and GraphQL compare, and when would you choose one over the others?**
* *Concepts:* REST (over-fetching/under-fetching, HTTP verbs), gRPC (HTTP/2, Protobufs, low-latency inter-service communication), GraphQL (single endpoint, client-defined schemas).


* **3. How do you handle API idempotency, especially for payment or order creation endpoints?**
* *Concepts:* `Idempotency-Key` headers, distributed locks (Redis), atomic database constraints, and preventing double-spend/duplicate entries.


* **4. What are the core strategies for API Rate Limiting, and how do you implement them at scale?**
* *Concepts:* Token Bucket, Leaky Bucket, Sliding Window Log algorithm, distributed caching (Redis counters), and API Gateways.


* **5. How do stateful sessions, JWTs, and OAuth2/OIDC differ for authentication and authorization?**
* *Concepts:* Session cookies vs. stateless JWTs (token revocation issues, expiration, signature verification), OAuth2 flows, and authorization models (RBAC/ABAC).



---

### 2. Databases & Storage Mechanics

* **6. How do B-Trees and B+ Trees work, and why are they used for database indexing instead of Binary Search Trees or Hash Tables?**
* *Concepts:* Disk I/O reduction, block/page storage, sequential read performance, and range query efficiency in relational databases.


* **7. What are the ACID properties, and how do database isolation levels balance correctness vs. performance?**
* *Concepts:* Atomicity, Consistency, Isolation, Durability. Isolation levels (Read Uncommitted, Read Committed, Repeatable Read, Serializable) and concurrency anomalies (Dirty Read, Non-repeatable Read, Phantom Read).


* **8. What is the CAP Theorem, and how does PACELC expand upon it?**
* *Concepts:* Consistency vs. Availability during network Partitions (CAP); Else (when no partition exists), Latency vs. Consistency (PACELC).


* **9. How do you identify, optimize, and eliminate slow database queries?**
* *Concepts:* Query execution plans (`EXPLAIN ANALYZE`), avoiding `N+1` query problems, eliminating `FULL TABLE SCANS`, covering indexes, and composite index ordering.


* **10. When should you choose a Relational (SQL) database over a NoSQL database (Document, Key-Value, Columnar, Graph)?**
* *Concepts:* Structured vs. unstructured data, strict ACID transactional needs vs. horizontal scalability, write throughput, and flexible schemas.


* **11. How do Optimistic and Pessimistic locking work to handle concurrent database updates?**
* *Concepts:* Pessimistic (`SELECT ... FOR UPDATE` row locks) vs. Optimistic (version/timestamp checks), lock contention, and preventing race conditions.



---

### 3. Caching & Performance Optimization

* **12. What are the standard Cache Eviction Policies and Cache Invalidation Strategies?**
* *Concepts:* Eviction (LRU, LFU, FIFO); Invalidation patterns (Cache-Aside, Write-Through, Write-Behind, Read-Through).


* **13. How do you mitigate the Cache Avalanche, Cache Stampede (Thundering Herd), and Cache Penetration problems?**
* *Concepts:* TTL jitter, distributed locking/single-flight patterns, mutexes, storing null values for missing keys, and Bloom Filters.


* **14. How does Redis achieve high throughput despite operating on a single-threaded execution model for commands?**
* *Concepts:* Non-blocking I/O multiplexing (`epoll`), execution in memory, avoiding thread context switching overhead, and atomic operations.



---

### 4. Distributed Systems & Messaging Architecture

* **15. How do Message Queues (RabbitMQ/SQS) and Distributed Event Logs (Kafka) differ?**
* *Concepts:* Push vs. Pull models, transient message deletion vs. append-only immutable logs with offset retention, and worker queues vs. stream processing.


* **16. How do you guarantee "At-Least-Once", "At-Most-Once", and "Exactly-Once" message delivery?**
* *Concepts:* Producer acknowledgements, consumer offset commits, retry policies, dead-letter queues (DLQ), and consumer-side idempotent processing.


* **17. How does the Transactional Outbox Pattern solve the dual-write problem between a Database and a Message Broker?**
* *Concepts:* Atomic database commits, decoupling event emission from business logic, and Change Data Capture (CDC) / tailing transaction logs.


* **18. How do horizontal scaling strategies like Database Sharding and Consistent Hashing work?**
* *Concepts:* Partition keys, hotspots, rehashing problems on cluster resizing, and Consistent Hashing rings with virtual nodes.


* **19. What is the Distributed Saga Pattern, and how does it replace Two-Phase Commit (2PC) in Microservices?**
* *Concepts:* Choreography vs. Orchestration, compensating transactions, eventual consistency, and avoiding long-lived distributed database locks.



---

### 5. Concurrency & Execution Mechanics

* **20. What is the fundamental difference between a Process and a Thread?**
* *Concepts:* Shared memory space vs. isolated address spaces, context switching overhead, CPU scheduling, and Inter-Process Communication (IPC).


* **21. How do Concurrency and Parallelism differ, and how does Asynchronous Non-blocking I/O work?**
* *Concepts:* Task interleaving vs. simultaneous multi-core execution; OS event loops (`epoll`, `kqueue`) and yielding execution during thread I/O wait times.


* **22. What are Race Conditions, Deadlocks, and Starvation, and how do you prevent them?**
* *Concepts:* Mutexes, Semaphores, Atomic variables, lock hierarchy enforcement, and avoiding nested resource acquisition.



---

### 6. Observability, Security & Production Readiness

* **23. How do you design for High Availability and Fault Tolerance?**
* *Concepts:* Circuit Breakers (Resilience4j/Hystrix), retries with exponential backoff and jitter, graceful degradation, and load shedding.


* **24. What are the key metrics for monitoring a production backend system?**
* *Concepts:* The 4 Golden Signals (Latency, Traffic, Errors, Saturation), RED method (Rate, Errors, Duration), distributed tracing (Trace IDs, Span IDs), and structured logging.


* **25. How do you prevent OWASP Top 10 security vulnerabilities on the backend?**
* *Concepts:* SQL Injection (parameterized queries), Cross-Site Scripting (XSS), Cross-Site Request Forgery (CSRF), Broken Object Level Authorization (BOLA/IDOR), and input sanitization.


* **26. How do zero-downtime deployment strategies work?**
* *Concepts:* Blue-Green deployments, Canary releases, Rolling updates, and backward-compatible database schema migrations (Expand-Contract pattern).



---

### 7. Core System Design Scenarios

* **27. How would you design a Rate Limiter or URL Shortener (e.g., Bitly)?**
* *Concepts:* High-throughput reads/writes, unique hash generation (Base62 vs. pre-generated key pools), Redis caching, DB selection, and race conditions.


* **28. How would you design an Event-Driven Notification System or Web Crawler at scale?**
* *Concepts:* Decoupled message queues, worker pools, distributed deduplication, rate-limiting external targets, and prioritizing queues.

**Core Request & Routing**

* **Request Context & Deadlines:** Learn `context.Context` for propagation, request cancellation, and setting timeouts (`context.WithTimeout`) so slow upstream services don't freeze server goroutines.
* **Input Validation & Sanitization:** Validate struct payloads (using tags or custom validators) before processing to prevent bad data or injection attacks.
* **Idiomatic Error Handling:** Standardize error wrapping with `fmt.Errorf("%w", err)` and construct unified JSON error responses across your API instead of returning raw internal errors.

**Data Persistence & Transactions**

* **Database Connections & Pooling:** Manage database pools with `database/sql` (`SetMaxOpenConns`, `SetMaxIdleConns`), understand connection lifetimes, and prevent connection starvation.
* **Database Migrations:** Schema version control using tools like `golang-migrate` to execute up/down migration files deterministically across environments.
* **Database Transactions & Isolation:** Execute multi-step atomic operations using `db.BeginTx()`, handling rollback on errors and understanding read/write lock implications.
* **Data Access Patterns (Repository Pattern):** Decouple business logic from persistence code using Go interfaces to facilitate unit testing and database swapping.

**Caching & State Management**

* **In-Memory & Distributed Caching:** Implement Cache-Aside and Write-Through patterns using Redis (`go-redis`) for expensive queries and session storage.
* **Cache Invalidation & TTL:** Set Time-To-Live strategies and cache invalidation mechanics to avoid stale data issues.

**Security & Protection**

* **Rate Limiting & Throttling:** Implement Token Bucket or Leaky Bucket algorithms (using `golang.org/x/time/rate` or Redis) to protect endpoints from abuse and DoS attacks.
* **Secrets Management & Config:** Load environment variables securely into structured Go configs without hardcoding credentials in repository code.

**Asynchronous Processing & Communication**

* **Worker Pools & Background Jobs:** Utilize Go channels and goroutines to construct bounded worker pools for heavy, non-blocking background tasks (e.g., sending emails, processing uploads).
* **Message Queues & Event-Driven Architecture:** Integrate message brokers like RabbitMQ or NATS to decouple services and process asynchronous job queues safely.

**Observability & Operations**

* **Structured Logging:** Move past `log.Println` to structured JSON logging using Go's standard `log/slog` library, appending contextual key-value pairs (trace IDs, user IDs).
* **Distributed Tracing & Request Correlation:** Generate or propagate correlation IDs (`X-Request-ID`) across incoming requests and outgoing downstream calls to trace execution paths.
* **Metrics Collection:** Instrument HTTP server endpoints using Prometheus (`prometheus/client_golang`) to measure request latency (histograms), error rates, and concurrent connection counters.
* **Graceful Shutdown:** Capture OS signals (`SIGINT`, `SIGTERM`) to stop accepting new HTTP connections and wait for in-flight requests and background workers to clean up before process exit.

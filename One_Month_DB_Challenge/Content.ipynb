### **Capstone Project: High-Concurrency Payment Engine**

**Goal:** Build a production-ready, idempotent digital wallet payment service in Go that guarantees data consistency under concurrent load.

**Core Requirements:**

1. **Tech Stack:** Go (Gin/Fiber/Chi), PostgreSQL, `sqlc`, `golang-migrate`, Redis, Docker.
2. **Features:**
* **Wallet Transfers:** Atomically transfer funds between two user accounts.
* **Concurrency Guard:** Use **Pessimistic Locking** (`SELECT FOR UPDATE`) on wallet rows in Postgres to prevent race conditions during simultaneous transfer requests.
* **Idempotency Engine:** A Redis-backed idempotency layer that caches request hashes for 24 hours to block duplicate charges from API retries.
* **Audit Logging:** Every balance change must write a corresponding immutable record to a `ledger_entries` table within the same DB transaction.
* **Performance Tuning:** Optimize DB connection limits, add composite indexes, and verify performance using `EXPLAIN ANALYZE` and concurrent Go benchmark tests.

Writing queries makes a system work; understanding engineering trade-offs keeps it running reliably under real-world load.

### **Week 1: Mechanical Sympathy & Storage Layout**

* **Day 1: Disk vs. Memory Mechanics**
* *Concept:* Sequential vs. random I/O latency (nanoseconds vs. milliseconds), OS page cache, NVMe access patterns.
* *Trade-off:* In-memory speed vs. durable disk persistence.


* **Day 2: Page Architecture & Layout**
* *Concept:* How databases structure data on disk (typically 8KB pages). Slotted page structures, headers, tuple pointers, pointer arrays.
* *Trade-off:* Space utilization vs. update overhead (fragmentation).


* **Day 3: Write-Ahead Logging (WAL)**
* *Concept:* The append-only log. Write to sequential WAL first, sync to disk, defer random page writes.
* *Trade-off:* Write latency (wait for `fsync`) vs. data durability guarantees.


* **Day 4: Storage Engines (B-Tree vs. LSM-Trees)**
* *Concept:* B-Trees (in-place updates, fast reads) vs. Log-Structured Merge Trees (append-only SSTables, fast writes, compaction).
* *Trade-off:* Read amplification (LSM) vs. Write amplification (B-Tree).


* **Day 5: Access Paths & Query Planner**
* *Concept:* Cost-based optimizers (CBO), page estimate statistics, Sequential Scan vs. Index Scan vs. Bitmap Index Scan.
* *Trade-off:* Fast query execution vs. optimizer statistics gathering overhead.


* **Day 6: Index Mechanics**
* *Concept:* B-Tree internal nodes, leaf pages, composite key ordering (leftmost prefix), cover indexes (Index Only Scans).
* *Trade-off:* Fast read paths vs. write penalty (every index must update on write).


* **Day 7: Memory Allocation & Buffer Pool**
* *Concept:* Buffer pool management (`shared_buffers`), dirty pages, LRU/Clock eviction algorithms, checkpointing.
* *Trade-off:* RAM allocation for query work (`work_mem`) vs. global page caching.

### **Week 2: Concurrency, MVCC & Data Isolation**

* **Day 8: Isolation Anomalies**
* *Concept:* Dirty Reads, Non-Repeatable Reads, Phantom Reads, Serialization Anomalies.
* *Trade-off:* Data correctness/strict consistency vs. transaction throughput.


* **Day 9: Multi-Version Concurrency Control (MVCC)**
* *Concept:* Tuple versioning via `xmin`/`xmax` system columns. Readers never block writers, writers never block readers.
* *Trade-off:* High concurrency vs. tuple bloat/garbage generation.


* **Day 10: Table Bloat & Housekeeping**
* *Concept:* Dead tuples, visibility maps, `VACUUM`, `AUTOVACUUM`, `HOT` (Heap-Only Tuple) updates.
* *Trade-off:* Background cleanup CPU/IO utilization vs. table bloat degradation.


* **Day 11: Locking Hierarchies**
* *Concept:* Row locks, table locks, advisory locks, lock queues, lock escalation.
* *Trade-off:* Fine-grained locking overhead vs. coarse-grained contention.


* **Day 12: Pessimistic Concurrency & Row Selection**
* *Concept:* `SELECT ... FOR UPDATE` vs. `FOR SHARE` vs. `SKIP LOCKED`.
* *Trade-off:* Blocking threads for safety vs. non-blocking skip execution for queue processing.


* **Day 13: Optimistic Concurrency Control (OCC)**
* *Concept:* CAS (Compare-And-Swap) patterns using row versions/timestamps without locks.
* *Trade-off:* Zero lock overhead vs. high retry overhead during heavy write contention.


* **Day 14: Deadlocks & Recovery**
* *Concept:* Lock wait graphs, cycle detection, automatic rollback strategies.
* *Trade-off:* Fail-fast rollbacks vs. long queue waiting.

### **Week 3: Connections, Drivers & Application Interface**

* **Day 15: Database Process & Connection Overhead**
* *Concept:* Connection cost (forking vs. thread per connection), memory allocated per client.
* *Trade-off:* Dedicated process safety vs. concurrency limit exhaustion.


* **Day 16: Connection Pooling Architecture**
* *Concept:* Client pools vs. Proxy pools (PgBouncer), Session vs. Transaction vs. Statement pooling mode.
* *Trade-off:* Stateful DB features (prepared statements, temporary tables) vs. massive connection scaling.


* **Day 17: Prepared Statements & Query Compilation**
* *Concept:* Parse -> Analyze -> Plan -> Execute pipeline. Query parameter binding vs. string interpolation.
* *Trade-off:* Statement cache memory usage vs. repeated query parsing CPU save.


* **Day 18: Object-Relational Impedance Mismatch**
* *Concept:* ORMs vs. Compile-time SQL generators (`sqlc`) vs. Raw SQL. N+1 queries.
* *Trade-off:* Developer speed/abstraction safety vs. query performance/predictability.


* **Day 19: Idempotency Mechanics**
* *Concept:* Idempotency keys, unique constraints, atomic `ON CONFLICT DO UPDATE/NOTHING` (UPSERT).
* *Trade-off:* Client-driven retries vs. server-side record deduplication.


* **Day 20: Migration Dynamics & Zero-Downtime Schema Changes**
* *Concept:* Non-blocking migrations, schema versioning, safe index creation (`CREATE INDEX CONCURRENTLY`).
* *Trade-off:* Multi-step deployment complexity vs. table lock outages during schema changes.


* **Day 21: Bulk Operations & Batching**
* *Concept:* Copy protocols (`COPY`), bulk insert batching, multi-row `VALUES` limits.
* *Trade-off:* Memory spike risks during large batches vs. round-trip latency overhead.

### **Week 22: Caching, Distributed Data & Resiliency**

* **Day 22: The In-Memory Paradigm**
* *Concept:* RAM data structures (Skiplists, ZipLists, Radix Trees), single-threaded event loop models (Redis).
* *Trade-off:* Microsecond latency vs. volatile storage & RAM cost constraints.


* **Day 23: Cache Coherency & Invalidation Patterns**
* *Concept:* Cache-Aside, Write-Through, Write-Back, Read-Through.
* *Trade-off:* Eventual consistency / stale reads vs. write path latency.


* **Day 24: Cache Failure Modes**
* *Concept:* Cache Stampede (Thundering Herd), Cache Penetration, Cache Avalanche, Mutex/Probabilistic early expiration solutions.
* *Trade-off:* Complex fallback mechanisms vs. database collapse during cache misses.


* **Day 25: Distributed Locking Concepts**
* *Concept:* Redis Redlock algorithm, Fencing tokens, TTL lock safety.
* *Trade-off:* Distributed lock complexity vs. duplicate execution risks in microservices.


* **Day 26: Distributed Data Foundations (CAP & PACELC)**
* *Concept:* Consistent Hashing, Primary-Replica Replication, Synchronous vs. Asynchronous Replication, Read Replicas.
* *Trade-off:* Read scalability vs. replication lag (stale reads).


* **Day 27: Partitioning & Sharding**
* *Concept:* Range, List, Hash Partitioning. Sharding keys, cross-shard query limits.
* *Trade-off:* Horizontal write scale vs. cross-shard transactional safety & complex joins.


* **Day 28: Reliability: Fault Tolerance & Rate Limiting**
* *Concept:* Leaky Bucket / Token Bucket algorithms, Connection timeouts, Circuit Breakers.
* *Trade-off:* Rejecting legitimate traffic vs. total system degradation under load.


* **Day 29: System Observability**
* *Concept:* Slow query logs, `pg_stat_activity`, `pg_stat_statements`, buffer cache hit ratios, lock profiling.
* *Trade-off:* Profiling/telemetry overhead vs. system black-box operational risks.


* **Day 30: System Synthesis & Trade-off Auditing**
* *Concept:* Evaluate complete systems end-to-end: Trade-offs between Consistency, Availability, Latency, Throughput, and Financial Cost.

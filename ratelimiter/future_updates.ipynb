## Highest leverage (do these first)

**1. Observability — Prometheus + Grafana**
Right now if this fell over in prod, you'd have no idea until someone complained. Add a `/metrics` endpoint (prometheus client_golang), expose request rate, 429 rate, p50/p95/p99 latency, Redis error rate, and collector buffer depth. Wire a Grafana dashboard on top. This is the single highest-signal addition — it's the difference between "built a rate limiter" and "built a rate limiter I could actually run on-call for." Every fintech interviewer you listed (Juspay, CRED, Razorpay) will probe on this.

**2. Chaos test: kill Redis mid-load**
You already fail open in the middleware. Prove it. Run vegeta while you `docker stop` the Redis container mid-run, capture that requests degrade to 503 instead of cascading failure, then bring Redis back and show recovery. This is a 30-minute add that gives you a genuinely good interview story ("I killed my dependency under load and verified graceful degradation") — most candidates at this level have never tested a failure path.

**3. A real README with a benchmark table**
This is what actually gets read before anyone opens your code. Naive version vs optimized version, real vegeta numbers (RPS sustained, p99 latency, memory), an architecture diagram (you already have the ASCII one — polish it), and the chaos test result. Quantify everything. This single file often decides whether a recruiter even clicks into the repo.

## Medium leverage

**4. Multiple rate-limiting algorithms, pluggable**
You've built sliding window. Add token bucket and fixed window as alternate `Limiter` implementations behind the same interface, and a short doc comparing their tradeoffs (burst tolerance, memory, accuracy). Shows you understand *why* sliding window was the right call, not just that you copied one algorithm.

**5. Policy management API**
CRUD endpoints (`POST/GET/PUT /policies/{client_id}`) backed by `client_policies` table you already have. Adds auth (a simple API key or JWT check) so it's not a wide-open admin surface. Turns this from "hardcoded limiter" into "a service other teams could actually configure."

**6. CI pipeline (GitHub Actions)**
`go test -race`, `go vet`, `golangci-lint`, on every push. Costs you 20 minutes, and a green Actions badge on your README is free credibility — it signals you don't just run tests locally when you remember to.

## Lower leverage but nice polish

**7. Deploy it somewhere live** (Fly.io / Railway, free tier) with a demo endpoint in the README. Concrete > theoretical for a non-brand-name resume.
**8. OpenTelemetry tracing** if you want to flex distributed-systems vocabulary — honestly optional unless you have time to spare; Prometheus metrics alone cover most of the signal.
**9. Redis Cluster** for HA — matches the stack you'd noted wanting (Redis Cluster, gRPC) for this project, but it's a bigger lift for lower marginal signal than 1–3 above. I'd only do this if you have days left over.

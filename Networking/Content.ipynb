Here is a structured 30-day practical learning roadmap designed for a backend engineer. Each day balances core theory with a concrete, hands-on task to reinforce how the concept works in code or at the system level.

### Phase 1: Layer 7 & HTTP Internals (Days 1–5)

* **Day 1: HTTP/1.1 Protocol Basics**
* **Concept:** HTTP request/response structure, headers, status codes, and persistent connections (`Keep-Alive`).
* **Mini-Practical:** Use `nc` (netcat) or `telnet` to write a raw HTTP/1.1 GET request to `example.com:80` and inspect the raw text response headers.


* **Day 2: Payload Formats & Serialization**
* **Concept:** JSON vs Protocol Buffers serialization overhead, binary framing, and schema enforcement.
* **Mini-Practical:** Write a benchmark script comparing serialization/deserialization speed and payload size of a struct using standard JSON vs Protocol Buffers.


* **Day 3: HTTP/2 Multiplexing & Framing**
* **Concept:** Binary framing layer, streams, multiplexing over a single TCP connection, and head-of-line blocking at the application layer.
* **Mini-Practical:** Run `curl -v --http2 [https://nghttp2.org](https://nghttp2.org)` to observe frame exchanges, multiplexed streams, and header compression (HPACK).


* **Day 4: WebSockets & Bidirectional Streaming**
* **Concept:** HTTP upgrade handshake, persistent TCP socket retention, framing, and heartbeat ping/pong mechanisms.
* **Mini-Practical:** Build a minimal WebSocket echo server and client to trace the initial `101 Switching Protocols` handshake and framing.


* **Day 5: gRPC & HTTP/3 (QUIC)**
* **Concept:** gRPC service definitions, streaming modes, and how HTTP/3 replaces TCP with UDP-based QUIC to solve network-level head-of-line blocking.
* **Mini-Practical:** Implement a simple unary gRPC service and inspect the traffic or logs to trace request/response metadata headers.



---

### Phase 2: Transport Layer & OS Sockets (Days 6–10)

* **Day 6: TCP Handshake & Connection Teardown**
* **Concept:** 3-way handshake (`SYN`, `SYN-ACK`, `ACK`), 4-way FIN teardown, and sequence/acknowledgment numbers.
* **Mini-Practical:** Start a Wireshark or `tcpdump` capture, trigger an HTTP call, and filter by `tcp.flags.syn == 1` to trace the 3-way handshake.


* **Day 7: TCP States & Connection Leaks**
* **Concept:** `TIME_WAIT`, `CLOSE_WAIT`, `FIN_WAIT_2`, socket exhaustion, and port allocation limits.
* **Mini-Practical:** Write a client script that opens hundreds of HTTP connections without closing them, then run `netstat -an | grep CLOSE_WAIT` or `ss -s` to observe state accumulation.


* **Day 8: TCP Performance Tuning**
* **Concept:** Nagle's algorithm (`TCP_NODELAY`), TCP window scaling, and keep-alives vs application-level heartbeats.
* **Mini-Practical:** Write a small TCP client sending small packets; toggle `TCP_NODELAY` on and off while capturing latency metrics.


* **Day 9: OS Sockets & File Descriptors**
* **Concept:** Socket syscalls (`socket`, `bind`, `listen`, `accept`), non-blocking sockets, and OS file descriptor limits (`ulimit`).
* **Mini-Practical:** Create a socket in your backend language, set non-blocking mode, and inspect the opened file descriptor in `/proc/<pid>/fd/`.


* **Day 10: I/O Multiplexing (`epoll` / `kqueue`)**
* **Concept:** Blocking I/O vs Thread-per-connection vs Event-driven I/O (`epoll` on Linux, `kqueue` on macOS).
* **Mini-Practical:** Explore how your language runtime implements its event loop or multiplexer (e.g., Go's netpoller, Node's `libuv`, or Python's `asyncio`).



---

### Phase 3: Security & Encryption (Days 11–14)

* **Day 11: Symmetric vs Asymmetric Encryption & PKI**
* **Concept:** Public/private keys, Certificate Authorities (CA), X.509 certificates, and trust chains.
* **Mini-Practical:** Generate a self-signed root CA and a server certificate using `openssl`.


* **Day 12: TLS Handshake & SNI**
* **Concept:** TLS 1.2 vs 1.3 handshake, Diffie-Hellman key exchange, Server Name Indication (SNI), and session resumption.
* **Mini-Practical:** Run `openssl s_client -connect google.com:443 -tls1_3` to inspect the full certificate chain and cipher negotiation.


* **Day 13: Mutual TLS (mTLS)**
* **Concept:** Zero-trust service-to-service authentication, client certificate validation, and identity propagation.
* **Mini-Practical:** Configure an HTTPS server to require client certificates and test connections using `curl --cert` vs raw `curl`.


* **Day 14: CORS, Security Headers & TLS Termination**
* **Concept:** Preflight requests (`OPTIONS`), `Access-Control-Allow-Origin`, HSTS, and edge vs downstream TLS termination.
* **Mini-Practical:** Configure NGINX or a lightweight gateway to handle TLS termination on port 443 and proxy plain HTTP traffic to a local backend on port 8000.



---

### Phase 4: DNS, Addressing & Routing (Days 15–19)

* **Day 15: IP Addressing & Subnetting**
* **Concept:** IPv4 vs IPv6, CIDR blocks (e.g., `/16`, `/24`), private IP ranges (`10.0.0.0/8`, `192.168.0.0/16`).
* **Mini-Practical:** Calculate host ranges for `172.16.0.0/22` and create two custom network interfaces locally to test subnets.


* **Day 16: DNS Lookup Architecture**
* **Concept:** Recursive resolvers, authoritative name servers, root servers, and record types (`A`, `AAAA`, `CNAME`, `MX`, `TXT`, `SRV`).
* **Mini-Practical:** Use `dig +trace example.com` to trace a DNS query from the root servers down to the authoritative name server.


* **Day 17: DNS Caching & TTL Issues**
* **Concept:** Local DNS caching, TTL propagation, stale record issues during blue-green deployments, and OS vs runtime resolution.
* **Mini-Practical:** Inspect local DNS resolver cache using `systemd-resolve --statistics` or `dscacheutil -statistics`, and test lookup behavior with short TTLs.


* **Day 18: Routing & Gateways**
* **Concept:** Routing tables, default gateways, packet forwarding, hop counts, and ICMP.
* **Mini-Practical:** Execute `traceroute` (or `mtr`) to a remote host and analyze your OS routing table via `netstat -rn` or `ip route`.


* **Day 19: Network Address Translation (NAT)**
* **Concept:** SNAT, DNAT, Port Address Translation (PAT), public vs private IP boundaries.
* **Mini-Practical:** Configure a local Docker bridge network and inspect how `iptables` rules handle NAT for container outbound traffic.



---

### Phase 5: Load Balancing & Proxies (Days 20–25)

* **Day 20: Reverse Proxies vs Forward Proxies**
* **Concept:** Proxy architecture, masking backends, header preservation (`X-Forwarded-For`, `X-Real-IP`).
* **Mini-Practical:** Set up NGINX or Caddy as a reverse proxy, pass requests to a simple HTTP application, and log the original client IP headers.


* **Day 21: Layer 4 vs Layer 7 Load Balancing**
* **Concept:** Packet-level routing (TCP/UDP) vs application-aware routing (HTTP headers, cookies, paths).
* **Mini-Practical:** Configure HAProxy or NGINX to load-balance traffic between two backend instances using round-robin.


* **Day 22: Consistent Hashing & Sticky Sessions**
* **Concept:** Session affinity, consistent hashing rings for cache distribution, and handling node additions/removals.
* **Mini-Practical:** Write a simple consistent hashing algorithm script to map 100 cache keys across 4 nodes and test remapping when 1 node drops.


* **Day 23: Connection Pooling & Socket Exhaustion**
* **Concept:** Reusing database/HTTP connections, idle timeouts, pool limits, and ephemeral port depletion.
* **Mini-Practical:** Build an HTTP client hitting a server with high concurrency; observe latency differences with connection pooling enabled vs disabled.


* **Day 24: Rate Limiting Algorithms**
* **Concept:** Token Bucket, Leaky Bucket, Fixed Window, Sliding Window Log/Counter.
* **Mini-Practical:** Implement a Redis-backed Sliding Window Rate Limiter middleware in your preferred backend language.


* **Day 25: Resilience Patterns: Circuit Breaking & Backpressure**
* **Concept:** Cascading failures, timeout strategies, retry storms with exponential backoff and jitter, and circuit breakers.
* **Mini-Practical:** Build a proxy client with exponential backoff + jitter logic to retry requests against a server returning `503 Service Unavailable`.



---

### Phase 6: Cloud Networking & Microservices (Days 26–30)

* **Day 26: Virtual Private Clouds (VPC) & Infrastructure Security**
* **Concept:** Public vs private subnets, NAT Gateways, Security Groups (stateful) vs Network ACLs (stateless).
* **Mini-Practical:** Use Terraform or AWS/GCP console to build a VPC with 1 public subnet, 1 private subnet, and a NAT gateway.


* **Day 27: Container Networking (Docker)**
* **Concept:** Docker network drivers (`bridge`, `host`, `none`, `overlay`), veth pairs, and network namespaces.
* **Mini-Practical:** Inspect network namespaces using `ip netns` or `docker inspect` to see how veth pairs link containers to the host bridge.


* **Day 28: Kubernetes Networking Basics**
* **Concept:** Pod-to-Pod communication model, ClusterIP, NodePort, LoadBalancer services, and CNI plugins.
* **Mini-Practical:** Spin up a local `minikube` or `k3d` cluster, deploy an app, and inspect DNS resolution between pods across namespaces.


* **Day 29: Ingress Controllers & API Gateways**
* **Concept:** Kubernetes Ingress, Envoy, path-based routing, authentication offloading, and rate limiting at the edge.
* **Mini-Practical:** Set up an NGINX or Traefik Ingress controller in your local cluster to route `/api/v1` and `/api/v2` to different backend services.


* **Day 30: Distributed Tracing & Service Mesh Basics**
* **Concept:** Sidecar proxy pattern (Envoy), mTLS auto-injection, and context propagation using W3C Trace Context headers (`traceparent`).
* **Mini-Practical:** Instrument a small 2-microservice call chain to pass and log the `traceparent` header across HTTP boundaries.

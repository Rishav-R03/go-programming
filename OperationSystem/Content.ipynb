Operating Systems provide the execution environment for your backend services. Understanding memory allocation, process scheduling, concurrency, and storage engines allows you to write high-throughput code, eliminate resource leaks, and diagnose production incidents.

Here is a 30-day OS roadmap focused purely on backend engineering fundamentals:

### Phase 1: Processes, Threads & Execution (Days 1–6)

* **Day 1: Process Abstraction & Anatomy**
* **Concept:** Virtual memory layout of a process (Stack, Heap, Data, BSS, Text), process state transitions (Running, Ready, Blocked), and Process Control Blocks (PCB).


* **Day 2: System Calls & Kernel vs User Mode**
* **Concept:** Ring 0 vs Ring 3 privilege levels, CPU trap/interrupt handling, context switching overhead, and common syscalls (`fork`, `execve`, `brk`, `read`, `write`).


* **Day 3: Threads & Concurrency Models**
* **Concept:** Kernel threads vs user-space threads (1:1, N:1, M:N models), OS thread scheduling, user/kernel context switches, and thread stack sizing.


* **Day 4: Process Creation & IPC Fundamentals**
* **Concept:** `fork()` copy-on-write (COW) mechanics, orphan/zombie processes, inter-process communication via Unix domain sockets, pipes, and shared memory.


* **Day 5: CPU Scheduling & Multithreading Overhead**
* **Concept:** Preemptive vs non-preemptive scheduling, Completely Fair Scheduler (CFS), CPU affinity, cache thrashing, and thread contention.


* **Day 6: Signal Handling & Process Lifecycle**
* **Concept:** POSIX signals (`SIGTERM`, `SIGKILL`, `SIGSEGV`, `SIGCHLD`), signal disposition, graceful shutdown patterns, and handling PID 1 in container environments.



---

### Phase 2: Memory Management & Garbage Collection (Days 7–12)

* **Day 7: Virtual Memory & Page Tables**
* **Concept:** Physical vs Virtual Address Space, Page Tables, Translation Lookaside Buffer (TLB), multi-level paging, and page sizes.


* **Day 8: Page Faults & Memory Allocation**
* **Concept:** Minor vs Major page faults, lazy memory allocation, `mmap()` vs `brk()`, and heap memory fragmentations.


* **Day 9: Memory Overcommit & The OOM Killer**
* **Concept:** Linux memory overcommit modes (`vm.overcommit_memory`), Out-Of-Memory (OOM) score calculation, swapped memory, and handling OOM events in microservices.


* **Day 10: User-Space Memory Allocators**
* **Concept:** How language allocators interact with the OS (`jemalloc`, `tcmalloc`, glibc `ptmalloc`), thread-local arenas, small/large object allocation, and fragmentation.


* **Day 11: Runtime Garbage Collection & OS Interactions**
* **Concept:** How language runtimes (Go/Java/V8) manage heap space atop OS virtual memory, STW (Stop-The-World) pauses, page commit/uncommit behaviors, and RSS vs Virtual Memory metrics.


* **Day 12: Shared Memory & Zero-Copy Architecture**
* **Concept:** Shared memory pages, POSIX shm, zero-copy buffer sharing, and eliminating user-kernel data copying.



---

### Phase 3: Concurrency, Synchronization & Locking (Days 13–17)

* **Day 13: Race Conditions & Critical Sections**
* **Concept:** Shared mutable state, atomic operations, hardware memory barriers, and compiler/CPU instruction reordering.


* **Day 14: Mutual Exclusion & Spinlocks**
* **Concept:** Mutexes vs Spinlocks, Test-And-Set instruction, OS sleep/wake primitives (`futex` on Linux), and lock acquisition latency.


* **Day 15: Deadlocks, Livelocks & Starvation**
* **Concept:** Coffman conditions for deadlocks, lock ordering strategies, deadlock detection, lock contention, and starvation in worker pools.


* **Day 16: Synchronization Primitives**
* **Concept:** Semaphores (counting vs binary), Condition Variables, Read-Write Locks (RWLocks), and Lock-Free data structures (CAS loop).


* **Day 17: Async I/O Runtimes & Non-Blocking Sockets**
* **Concept:** Reactive architecture vs thread-per-request model, OS readiness notifications (`epoll`/`kqueue`), completion-based I/O (`io_uring`), and green threads/coroutines.



---

### Phase 4: File Systems, Storage & Page Cache (Days 18–22)

* **Day 18: File System Architecture & Inodes**
* **Concept:** File Descriptors, Inode tables, directory structures, hard vs soft links, and block sizes.


* **Day 19: The OS Page Cache & Write Buffering**
* **Concept:** Page Cache architecture, dirty pages, write-back vs write-through caching, and kernel background flushing (`pdflush`/`flush` threads).


* **Day 20: I/O Operations & Sync Primitives**
* **Concept:** Sequential vs Random I/O throughput, `fsync()`, `fdatasync()`, `O_DIRECT` (bypassing page cache), and `O_SYNC` flags.


* **Day 21: File Locking & Concurrent File Access**
* **Concept:** Advisory vs Mandatory file locks, `flock()` vs `fcntl()`, multi-process file writing, and log file truncation dynamics.


* **Day 22: Storage Engines & OS File Systems**
* **Concept:** How databases (PostgreSQL, MySQL, Redis) use OS file primitives: Write-Ahead Logs (WAL), Append-Only Files (AOF), memory-mapped files (`mmap`), and SSD-friendly block allocation.



---

### Phase 5: Resource Control, Isolation & Virtualization (Days 23–27)

* **Day 23: Linux Namespaces & Isolation**
* **Concept:** PID, Mount, Network, IPC, UTS, and User namespaces—how the OS creates process isolation without virtualization.


* **Day 24: Control Groups (cgroups v1 & v2)**
* **Concept:** Resource limiting (CPU quotas, memory limits, I/O throttling, pids limit), CFS bandwidth control, and cgroup-driven OOM killing.


* **Day 25: Containers Internals (Docker/runc)**
* **Concept:** How containers use Namespaces, cgroups, and OverlayFS (union file systems) to build lightweight execution environments.


* **Day 26: Virtualization vs Containerization**
* **Concept:** Type-1 vs Type-2 Hypervisors, hardware-assisted virtualization (KVM, Intel VT-x), guest OS kernel vs host kernel sharing.


* **Day 27: Resource Limits & OS Tunables**
* **Concept:** POSIX `ulimit` settings (open file descriptors, max user processes, stack size) and `/etc/security/limits.conf`.



---

### Phase 6: OS Observability, Performance & Troubleshooting (Days 28–30)

* **Day 28: System Telemetry & Metrics**
* **Concept:** `/proc` and `/sys` pseudo-file systems, Load Average calculation (Uninterruptible Sleep state `D`), CPU states (user, system, iowait, steal).


* **Day 29: Low-Level Profiling & Tracing**
* **Concept:** Syscall tracing (`strace`), library call tracing (`ltrace`), CPU profiling (`perf`), and eBPF (Extended Berkeley Packet Filter) observability.


* **Day 30: Kernel Tuning for High-Throughput Backends**
* **Concept:** Key Sysctl tunables: `sysctl.conf`, TCP buffer sizing (`rmem`/`wmem`), socket backlog (`somaxconn`), ephemeral port range (`ip_local_port_range`), and swap aggressiveness (`swappiness`).

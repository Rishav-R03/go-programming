To show startup readiness as an SDE-1 backend engineer in Go, knowing syntax and passing DSA interviews isn't enough. Startups need engineers who understand **goroutine lifetimes, memory allocation overhead, failure domain boundaries, production observability, and how to write Idiomatic Go** without over-engineering.

This 30-day curriculum moves you from Go internal mechanics to distributed fault tolerance, culminating in an asynchronous background job processing engine.

### **Capstone Project: Asynchronous Task Processor (Asynq/Celery clone in Pure Go)**

**Goal:** Build a robust, concurrent, distributed background job processing engine in Go that handles background tasks safely without dropping jobs or leaking resources.

**Architecture Requirements:**

1. **Concurrency Control:** Fixed worker pools reading from Redis queues via atomic operations.
2. **Context Deadlines:** Enforce execution timeouts on individual jobs using `context.WithTimeout`.
3. **Resiliency:** Automatic retries with **exponential backoff + jitter** for failed tasks, moving permanently failing tasks to a **Dead Letter Queue (DLQ)**.
4. **Graceful Degradation:** Catch `SIGTERM` signals, allow running tasks to complete, and safely re-queue unstarted jobs back into Redis.
5. **Observability:** Integrate `log/slog` for structured tracing and expose a `/debug/pprof` endpoint to prove zero memory or goroutine leaks.
